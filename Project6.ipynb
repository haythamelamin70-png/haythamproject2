```python
import os
from google.colab import userdata
from crewai import Agent, Task, Crew, Process, LLM

# Configure OPENROUTER API Key
os.environ["OPENROUTER_API_KEY"] = userdata.get("OPENROUTER_API_KEY")

# Instantiate the CrewAI LLM wrapper for Groq
groq_llm = LLM(
    model="openrouter/openrouter/free",
    api_key=os.environ["OPENROUTER_API_KEY"]
)

# 1. Agent
researcher = Agent(
    role="AI Researcher",
    goal="Research the latest AI technology",
    backstory="You are an AI research expert",
    llm=groq_llm
)

# 2. Task

question = input("Enter your question: ")
research_task = Task(
    description=f"Answer this question clearly: {question}",
    expected_output="A simple summary of AI developments",
    agent=researcher
)

# 3. Crew
my_crew = Crew(
    agents=[researcher],
    tasks=[research_task],
    process=Process.sequential
)

# 4. Run
result = await my_crew.kickoff_async()

# 5. Show result
print(result)
```
