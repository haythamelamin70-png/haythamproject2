```python
# Importing class Agent, class Task, class Crew, Class Process from crewai liberary
from crewai import Agent, Task, Crew, Process


researcher= Agent(

  role="Senior research analyst",
  goal="Discover ground breaking technologies in SCADA system ",
  backstory="An expert analyst with an knock for hidind gymmics in data "

                  )

writer= Agent(

  role="Tech contexts stratigist",
  goal="Craft compelling content on technical advacement",
  backstory="A skil writer refine complex topics into clear naritives"

              )

researcher_task= Task(

  description="Identify key in use in SCADA system 2026.",
  agent=researcher,
  expected_output="A comprehensive report on key AI trends and applications in SCADA system sector for 2026."
                      )



writer_task= Task(

 description="Compose three pagraph summarising the key trend foumd.",
 agent=writer,
 expected_output="A three-paragraph summary, well-written and engaging, highlighting the most significant trends found by the researcher."
                  )
```

```python


research_crew=Crew(

                    agents=[researcher, writer],
                    tasks=[researcher_task, writer_task],
                    process=Process.sequential
)

```
