```python

# 1. Import the required libraries
import os
from google.colab import userdata
from crewai import Agent, Task, Crew, Process, LLM

# Configure OPENROUTER API Key
os.environ["OPENROUTER_API_KEY"] = userdata.get("OPENROUTER_API_KEY")

# Instantiate the CrewAI LLM wrapper for Groq_llm virable. groq_llm is just a variable name, not Groq API.
groq_llm = LLM(
    model="openrouter/openrouter/free",
    api_key=os.environ["OPENROUTER_API_KEY"]
)


# 2. Create the Researcher Agent (using the custom LLM)
researcher = Agent(
    role="Senior Research Analyst",
    goal="Discover groundbreaking technologies in SCADA systems",
    backstory="An expert analyst with a knack for finding hidden patterns in data.",
    llm=groq_llm
)

# 3. Create the Writer Agent (using the custom LLM)
writer = Agent(
    role="Technical Content Strategist",
    goal="Create compelling content about technological advancements",
    backstory="A skilled writer who explains complex topics clearly.",
    llm=groq_llm
)

# 4. Create the Research Task
researcher_task = Task(
    description="Identify key AI technologies and trends used in SCADA systems in 2026.",
    expected_output="A comprehensive report on key AI in SCADA systems in 2026.",
    agent=researcher
)


# 5. Create the Writing Task
writer_task = Task(
    description="Write three paragraphs summarising the key trends.",
    expected_output="Three clear and engaging paragraphs highlighting the most significant SCADA AI trends.",
    agent=writer
)


# 6. Create the Crew
research_crew = Crew(
    agents=[researcher, writer],
    tasks=[researcher_task, writer_task],
    process=Process.sequential
)

# 7. Run the Crew asynchronously using await in the Colab notebook
result = await research_crew.kickoff_async()

# 8. Display the final result
print("###############")

print(result)
```
