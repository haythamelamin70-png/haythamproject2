```python
import os
from google.colab import userdata
from crewai import Agent, Task, Crew, Process, LLM

# Configure OPENROUTER API Key
os.environ["OPENROUTER_API_KEY"] = userdata.get("OPENROUTER_API_KEY")

# Instantiate the CrewAI LLM wrapper for My_ll
My_llm = LLM(
    model="openrouter/openrouter/free",
    api_key=os.environ["OPENROUTER_API_KEY"]
)

# 1. Agent1 researcher
researcher = Agent(
    role="AI Researcher",
    goal="Research the latest AI technology",
    backstory="You are an AI research expert",
    llm=My_llm
)

# 2. Agent2  writer

writer = Agent(
    role="AI Writer",
    goal="Write a blog post",
    backstory="You are an AI writer",
    llm=My_llm
)

# 3. Task1 researcher_task

question = input("Enter your question: ")
research_task = Task(
    description=f"Answer this question clearly: {question}",
    expected_output="A simple summary of AI developments",
    agent=researcher
)

# 4. Task2 writer_task

writer_task = Task(
    description="Write a brief, clear, simple post based on the researcher's answer.",
    expected_output="A brief, clear, simple post about AI developments.",
    agent=writer,
    context=[research_task] # context is specified to question
)


# 5. Crew
my_crew = Crew(
    agents=[researcher,writer],
    tasks=[research_task, writer_task],
    process=Process.sequential
)

# 6. Run
result = await my_crew.kickoff_async()

# 7. Remove Markdown stars and hashes by importin class re
import re
clean_result = re.sub(r'[#*]', '', str(result))


# 8. Show clean result
print(clean_result)
```
